```python
import pandas as pd
import numpy as np
df=pd.read_csv("country_wise_latest.csv")
print(df.head())
```

```python
from scipy import stats

print(df.head())
print(df.describe())

mean_val = df['Deaths'].mean()
median_val = df['Deaths'].median()
mode_val = df['Deaths'].mode()[0]
print('Mean:', mean_val)
print('Median:', median_val)
print('Mode:', mode_val)


Deaths_array = df['Deaths'].to_numpy()
print('NumPy Mean:', np.mean(Deaths_array))
print('NumPy Median:', np.median(Deaths_array))
print('SciPy Mode:', stats.mode(Deaths_array, keepdims=True))

group_summary = df.groupby('Country/Region')['Deaths'].agg(['mean', 'median',lambda x: x.mode()[0]])
group_summary.columns = ['Mean', 'Median', 'Mode']
print(group_summary)

print('Skewness:', df['Deaths'].skew())
if mean_val > median_val:
     print('Distribution is right (positively) skewed')
elif mean_val < median_val:
     print('Distribution is left (negatively) skewed')
else:
     print('Distribution is approximately symmetric')
```
