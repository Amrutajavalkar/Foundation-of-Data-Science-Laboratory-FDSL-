```python
import pandas as pd
import numpy as np
df=pd.read_csv("credit_card_fraud_2026.csv")
# Total amount used by card
print("Sum of amount by card_type:")
display(df.groupby("card_type")["amount_usd"].sum())

# Average price per card
print("\nMean amount by card type:")
display(df.groupby("card_type")["amount_usd"].mean())

# Multiple aggregations at once with agg()
print("\nMultiple aggregations with agg():")
display(df.groupby("card_type").agg(
    total_qty=("amount_usd", "sum"),
    avg_price=("hours_since_last_txn", "mean"),
    order_count=("transaction_id", "count")
))
```

```python
pivot = pd.pivot_table(
    df,
    values="amount_usd",
    index="card_type",
    columns="device_type",
    aggfunc="sum",
    fill_value=0
)
pivot
```

```python
print("Missing values per column:")
display(df.isnull().sum())

print("\nRows with any missing values:")
display(df[df.isnull().any(axis=1)])

print("\nDropping rows with missing values (dropna):")
display(df.dropna())

print("\nFilling missing Quantity with the column mean (fillna):")
df_filled = df.copy()
df_filled["amount_usd"] = df_filled["amount_usd"].fillna(df_filled["amount_usd"].mean())
display(df_filled)
```

```python
df_t = df_filled.copy()

# apply(): create a new column using a custom function
df_t["Total"] = df_t.apply(lambda row: row["amount_usd"] * row["hours_since_last_txn"], axis=1)

# map(): map values in a column to new values
card_ttype = {"Visa": "Visa", "Mastercard": "Mastercard", "Discover": "Discover", "Amex": "Amex"}
df_t["card_type"] = df_t["card_type"].map(card_ttype)

# astype(): change a column's data type
df_t["amount_usd"] = df_t["amount_usd"].astype(int)

# rename(): rename columns
df_t = df_t.rename(columns={"Total": "TotalRevenue"})

# sort_values(): sort rows by a column
df_t = df_t.sort_values(by="transaction_id", ascending=False)

df_t
```
