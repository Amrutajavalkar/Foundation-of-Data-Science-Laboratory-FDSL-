```python
import numpy as np
import pandas as pd
df=pd.read_csv('country_wise_latest.csv')

#Top 5 Rows
print(df.head())
# To find out the range of Deaths and Active contries
range_death=df['Deaths'].max()-df['Deaths'].min()
range_active=df['Active'].max()-df['Active'].min()

print(range_death)
print(range_active)

mean=df['Deaths'].mean()
print(mean)
varience=(df['Deaths']).var()
print(varience)

sd=np.sqrt(varience)
print(sd)


#regression coefficient
reg=(sd/mean)*100
print(reg)
```
